##**Loading the raw data**

```python
import pandas as pd
df = pd.read_csv('noshow_appointments.csv')
df.head()
```

##**Cleaning the data (this is the evidence behind Section 5 — data quality issues)**

```python
# Fix column name typos and clean up
df = df.rename(columns={'Hipertension': 'Hypertension', 'Handcap': 'Handicap', 'No-show': 'NoShow'})

# Convert dates
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])

# Feature: wait time in days (this is a KEY predictive feature — available at prediction time)
df['WaitDays'] = (df['AppointmentDay'] - df['ScheduledDay']).dt.days

# Drop invalid age
df = df[df['Age'] >= 0]

# Handle negative wait days (data entry artifacts) — clip at 0
df['WaitDays'] = df['WaitDays'].clip(lower=0)

# Convert target to binary: 1 = No-show, 0 = Showed up
df['NoShow'] = df['NoShow'].map({'Yes': 1, 'No': 0})

print(df.shape)
print(df['NoShow'].value_counts(normalize=True))
df.head()
```

##**Exploratory analysis (also supports Section 5)**

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Wait days vs no-show rate
df['WaitBucket'] = pd.cut(df['WaitDays'], bins=[-1,0,3,7,14,30,180], labels=['0','1-3','4-7','8-14','15-30','30+'])
df.groupby('WaitBucket')['NoShow'].mean().plot(kind='bar', ax=axes[0], title='No-show rate by wait time')

# SMS received vs no-show
df.groupby('SMS_received')['NoShow'].mean().plot(kind='bar', ax=axes[1], title='No-show rate by SMS reminder')

# Age vs no-show
df['AgeBucket'] = pd.cut(df['Age'], bins=[0,12,18,30,45,60,120], labels=['0-12','13-18','19-30','31-45','46-60','60+'])
df.groupby('AgeBucket')['NoShow'].mean().plot(kind='bar', ax=axes[2], title='No-show rate by age group')

plt.tight_layout()
plt.show()
```

##**Train/test split → answers Section 6, row "Data splitting/validation strategy"**

```python
from sklearn.model_selection import train_test_split

# Select features
features = ['Gender', 'Age', 'Scholarship', 'Hypertension', 'Diabetes',
            'Alcoholism', 'Handicap', 'SMS_received', 'WaitDays']
target = 'NoShow'

# One-hot encode Gender
df_model = pd.get_dummies(df[features + [target, 'PatientId']], columns=['Gender'], drop_first=True)

# Split by unique patient to avoid leakage across train/test
patient_ids = df_model['PatientId'].unique()
train_ids, test_ids = train_test_split(patient_ids, test_size=0.2, random_state=42)

train_df = df_model[df_model['PatientId'].isin(train_ids)]
test_df = df_model[df_model['PatientId'].isin(test_ids)]

X_train = train_df.drop(columns=[target, 'PatientId'])
y_train = train_df[target]
X_test = test_df.drop(columns=[target, 'PatientId'])
y_test = test_df[target]

print("Train shape:", X_train.shape, "Test shape:", X_test.shape)
print("Train no-show rate:", y_train.mean(), "Test no-show rate:", y_test.mean())
```

##**Baseline model → answers Section 6, row "Proposed baseline"**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score

# Train baseline
baseline = LogisticRegression(max_iter=1000, class_weight='balanced')
baseline.fit(X_train, y_train)

# Predict
y_pred = baseline.predict(X_test)
y_proba = baseline.predict_proba(X_test)[:, 1]

print("Classification report:")
print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
```

##**Main model → answers Section 6, row "Main modeling approach"**

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=200, max_depth=10, class_weight='balanced', random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
y_proba_rf = rf.predict_proba(X_test)[:, 1]

print("Random Forest classification report:")
print(classification_report(y_test, y_pred_rf))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_rf))
```

##**Feature importance → answers Section 6, row "Main technical risks/assumptions"**

```python
import pandas as pd

importances = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
print(importances)

importances.plot(kind='barh', figsize=(8,5), title='Feature Importance (Random Forest)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

##**Saving the model → answers Section 8, deliverable "trained model or ML pipeline"**

**Step 1 — Save the trained model:**

```python
import joblib
joblib.dump(rf, 'noshow_model.pkl')
print("Model saved")
```

##**Inference function → answers Section 6, rows "Expected inference input/output" and Section 8, deliverable "usable inference interface"**

**Step 2 — Build a simple inference function (takes a new patient/appointment record, returns a no-show risk prediction):**

```python
def predict_noshow(age, scholarship, hypertension, diabetes, alcoholism,
                    handicap, sms_received, wait_days, gender_m):
    input_data = pd.DataFrame([{
        'Age': age,
        'Scholarship': scholarship,
        'Hypertension': hypertension,
        'Diabetes': diabetes,
        'Alcoholism': alcoholism,
        'Handicap': handicap,
        'SMS_received': sms_received,
        'WaitDays': wait_days,
        'Gender_M': gender_m
    }])[X_train.columns]  # ensure same column order

    proba = rf.predict_proba(input_data)[0, 1]
    prediction = "Likely No-Show" if proba >= 0.5 else "Likely to Show"
    return prediction, round(proba, 3)

# Example: 35-year-old male, no conditions, SMS sent, booked 20 days ahead
result = predict_noshow(age=35, scholarship=0, hypertension=0, diabetes=0,
                          alcoholism=0, handicap=0, sms_received=1, wait_days=20, gender_m=1)
print(result)
```

##**Step 1 — Add the missing validation/confusion-matrix/reload-check code (paste as one new cell after your inference function cell):**

```python
# Input validation added to inference function
def predict_noshow(age, scholarship, hypertension, diabetes, alcoholism,
                    handicap, sms_received, wait_days, gender_m):
    if age < 0 or age > 115:
        return "Invalid input: age out of expected range", None
    if wait_days < 0:
        return "Invalid input: wait_days cannot be negative", None
    if any(v not in [0, 1] for v in [scholarship, hypertension, diabetes, alcoholism, sms_received, gender_m]):
        return "Invalid input: binary fields must be 0 or 1", None

    input_data = pd.DataFrame([{
        'Age': age, 'Scholarship': scholarship, 'Hypertension': hypertension,
        'Diabetes': diabetes, 'Alcoholism': alcoholism, 'Handicap': handicap,
        'SMS_received': sms_received, 'WaitDays': wait_days, 'Gender_M': gender_m
    }])[X_train.columns]

    proba = rf.predict_proba(input_data)[0, 1]
    prediction = "Likely No-Show" if proba >= 0.5 else "Likely to Show"
    return prediction, round(proba, 3)

# Confusion matrix
from sklearn.metrics import ConfusionMatrixDisplay
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_rf)
plt.title('Random Forest — Confusion Matrix')
plt.show()

# Reload check — proves the saved model file actually works standalone
loaded_model = joblib.load('noshow_model.pkl')
match = (loaded_model.predict_proba(X_test.iloc[[0]]) == rf.predict_proba(X_test.iloc[[0]])).all()
print("Reloaded model prediction matches original:", match)

# Test invalid input handling
print(predict_noshow(age=-5, scholarship=0, hypertension=0, diabetes=0, alcoholism=0, handicap=0, sms_received=1, wait_days=5, gender_m=1))
```

**Section 5 — Data & Problem Discovery**

| Question | Response |
|---|---|
| Selected dataset and source | Kaggle "Medical Appointment No Shows" dataset (Brazil, 2016), publicly available, 110,527 appointment records across 62,299 unique patients |
| What does one record represent | One scheduled medical appointment for a patient at a public health clinic |
| Proposed target/ML objective | Binary classification: predict `No-show` (Yes/No) for a scheduled appointment |
| Key info available at prediction time | Age, gender, neighbourhood, scheduling lead time (WaitDays), SMS reminder flag, chronic conditions, scholarship/welfare status — all known before the appointment occurs |
| Main data quality issues | One record had Age = -1 (removed); ~35% of records had negative wait-day values (clipped to 0) |
| Potential leakage risks | Low — all features are recorded before the outcome is known. Split was done at the patient level to prevent leakage |
| Privacy/fairness/licensing concerns | Dataset is de-identified and public. Neighbourhood may act as a socioeconomic proxy — flagged as a fairness consideration |
